```python
df["Date"] = pd.to_datetime(df["YEAR"] * 1000 + df["DOY"], format="%Y%j")
df["Month"] = df["Date"].dt.month
```

```python
monthly_temp = df.groupby("Month")["T2M"].mean()
monthly_temp
```

```python
import matplotlib.pyplot as plt

monthly_temp.plot(kind="line")
plt.title("Average Monthly Temperature (Ethiopia)")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
monthly_rain = df.groupby("Month")["PRECTOTCORR"].sum()
monthly_rain
```

```python
monthly_rain.plot(kind="bar")
plt.title("Monthly Rainfall (Ethiopia)")
plt.xlabel("Month")
plt.ylabel("Rainfall (mm)")
plt.show()
```

##Temperature and Rainfall Analysis (Ethiopia)

-The average temperature is highest around April–May 
-The lowest temperatures occur around November- December 
-Rainfall peaks during July-August, indicating the main rainy season 
-Months with very low rainfall represent the dry season, especially December-January

```python
import seaborn as sns

plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

### Correlation Analysis (Ethiopia)

- There is a strong positive correlation between **T2M and T2M_MIN (0.72)** and **T2M and T2M_MAX (0.65)**, indicating that average temperature increases with both minimum and maximum temperatures.

- **WS2M and WS2M_MAX (0.94)** show a very strong positive correlation, meaning higher average wind speeds are closely associated with higher maximum wind speeds.

- **RH2M and QV2M (0.90)** have a strong positive relationship, suggesting that higher relative humidity corresponds with higher specific humidity.

- A strong negative correlation exists between **T2M_RANGE and QV2M (-0.89)** and between **T2M_RANGE and RH2M (-0.87)**, indicating that larger temperature variations are associated with lower humidity levels.

- **T2M_MAX and RH2M (-0.79)** also show a strong negative relationship, suggesting that higher maximum temperatures tend to occur when humidity is lower.

- Precipitation (**PRECTOTCORR**) has moderate positive correlation with **RH2M (0.5)** and **QV2M (0.52)**, indicating rainfall is associated with higher moisture levels in the air.

- Most other variables show weak or negligible correlations, suggesting limited direct linear relationships.
